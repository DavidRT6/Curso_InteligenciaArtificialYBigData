# Enunciado del Proyecto en R: Análisis de Ventas y Gastos en una Empresa con 5 Tiendas

## Descripción
Una empresa cuenta con **5 tiendas**, y se desea analizar sus ventas y gastos mensuales durante **un año (12 meses)**. Los gastos están divididos en las siguientes categorías:
- **Material**
- **Personal**
- **Publicidad**
- **Gastos eléctricos**

Deberéis desarrollar un programa en **R** que permita analizar los datos generados de manera lógica (sin valores extremadamente dispares).

---

<h2><b><u>Objetivos del Programa</u></b></h2>

<h2><b><u>0. Preparación del análisis</u></b></h2>

En este apartado se va a mostrar el código necesario para poder realizar los análisis solicitados.

```python
# Librerías utilizadas
import numpy as np
import pandas as pd
from scipy.stats import mode
```

```python
# Configurar semilla para que los datos sean siempre los mismos
np.random.seed(123)

# Definir los nombres de las tiendas y los meses
tiendas = ["Tienda Sevilla", "Tienda Barcelona", "Tienda Valencia", "Tienda Baleares", "Tienda Santiago"]
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"]

# Generar datos aleatorios dentro de los rangos especificados
cantidad_datos_necesarios = len(meses) * len(tiendas)

ingresos = np.random.randint(50000, 150001, size=(len(tiendas), len(meses)))
gastos_material = np.random.randint(5000, 20001, size=(len(tiendas), len(meses)))
gastos_personal = np.random.randint(20000, 50001, size=(len(tiendas), len(meses)))
gastos_publicidad = np.random.randint(3000, 10001, size=(len(tiendas), len(meses)))
gastos_electricos = np.random.randint(2000, 7001, size=(len(tiendas), len(meses)))

total_gastos = gastos_material + gastos_personal + gastos_publicidad + gastos_electricos
beneficio = ingresos - total_gastos

# Convertir a DataFrames
df_ingresos = pd.DataFrame(ingresos, index=tiendas, columns=meses)
df_gastos = pd.DataFrame(total_gastos, index=tiendas, columns=meses)
df_beneficio = pd.DataFrame(beneficio, index=tiendas, columns=meses)
```

**Explicación:**

Este código genera datos aleatorios para simular los ingresos, gastos y beneficios de varias tiendas a lo largo de un año. A continuación, se explica de manera más detallada:

- **Configurar la semilla aleatoria "np.random.seed(123)"**: para garantizar que los valores aleatorios generados sean siempre los mismos en cada ejecución.

- **Definir los nombres de tiendas y meses**:
    - Tiendas: lista que contiene los nombres de cinco tiendas.
    - Meses: lista que contiene los nombres de los doce meses del año.

- **Calcular la cantidad de datos necesarios**: se multiplica la cantidad de meses "len(meses)" por la cantidad de tiendas "len(tiendas)" para determinar cuántos valores aleatorios se necesitan.

- **Generación de datos aleatorios**<br> 
Se crean arrays de numpy de valores aleatorios utilizando "np.random.randint" con dimensiones de 5 filas (tiendas) por 12 columnas (meses), con los siguientes rangos:
    - ingresos: entre 50.000 y 150.000.
    - gastos_material: entre 5.000 y 20.000.
    - gastos_personal: entre 20.000 y 50.000.
    - gastos_publicidad: entre 3.000 y 10.000.
    - gastos_electricos: entre 2.000 y 7.000.

- **Cálculo de gastos totales y beneficios**
    - total_gastos: suma de todos los gastos de cada tienda en cada mes.
    - beneficio: diferencia entre los ingresos y los gastos totales.

- **Convertir a DataFrames**<br>
Para facilitar el análisis, los datos generados se convierten en DataFrames de Pandas, donde:
    - Filas: representan las tiendas.
    - Columnas: representan los meses.

---

<h2><b><u>1. Cálculo de Totales</u></b></h2>

**A nivel global** (sumando todas las tiendas):
- Total de ingresos mensuales.
- Total de cada tipo de gasto mensual.
- Beneficio mensual total (beneficio = ingresos - gastos).

**A nivel de cada tienda**:
- Ingresos mensuales de cada tienda.
- Total de cada tipo de gasto mensual por tienda.
- Beneficio mensual por tienda.

```python
# 1. Cálculo de totales globales y de cada tienda
print("1. CALCULO DE TOTALES:")
print("Ingresos totales por mes:")
print(df_ingresos.sum())
print("Gastos totales por mes:")
print(df_gastos.sum())
print("Beneficio total por mes:")
print(df_beneficio.sum())

print("\nIngresos totales por tienda:")
print(df_ingresos.sum(axis=1))
print("Gastos totales por tienda:")
print(df_gastos.sum(axis=1))
print("Beneficio total por tienda:")
print(df_beneficio.sum(axis=1))
```

**Explicación**:
- `sum()`: suma las filas de un dataframe.
- `sum(axis=1)`: suma las columnas de un dataframe.

---

<h2><b><u>2. Análisis de los Meses Clave</u></b></h2>

**A nivel global**:
- Mes con mayores ingresos y mes con menores ingresos.
- Mes con mayor beneficio y mes con menor beneficio.

**A nivel de cada tienda**:
- Mes con mayores ingresos y mes con menores ingresos por tienda.
- Mes con mayor beneficio y mes con menor beneficio por tienda.

```python
# 2. Análisis de los meses clave
print("2. ANALISIS DE LOS MESES CLAVE:")
print("Mes con mayores ingresos:", df_ingresos.sum().idxmax())
print("Mes con menores ingresos:", df_ingresos.sum().idxmin())
print("Mes con mayor beneficio:", df_beneficio.sum().idxmax())
print("Mes con menor beneficio:", df_beneficio.sum().idxmin())

print("\nMes con mayores ingresos por tienda:")
print(df_ingresos.idxmax(axis=1))
print("Mes con menores ingresos por tienda:")
print(df_ingresos.idxmin(axis=1))
print("Mes con mayor beneficio por tienda:")
print(df_beneficio.idxmax(axis=1))
print("Mes con menor beneficio por tienda:")
print(df_beneficio.idxmin(axis=1))
```

**Explicación**
- `sum().idxmax()`: "sum()" suma las filas del dataframe e "idxmax()" obtiene el nombre de la columna con el valor máximo.
- `sum().idxmin()`: "sum()" suma las filas del dataframe e "idxmin()" obtiene el nombre de la columna con el valor mínimo.

- `idxmax(axis=1)`: obtiene el nombre de la columna, de la fila con el valor máximo.
- `idxmin(axis=1)`: obtiene el nombre de la columna, de la fila con el valor mínimo.

---

<h2><b><u>3. Estadísticas Mensuales</u></b></h2>

**A nivel global**:
- Media, mediana, moda y desviación típica de los ingresos mensuales.
- Media, mediana, moda y desviación típica del beneficio mensual.

**A nivel de cada tienda**:
- Media, mediana, moda y desviación típica de los ingresos por tienda.
- Media, mediana, moda y desviación típica del beneficio por tienda.

```python
# 3. Estadísticas mensuales
print("3. ESTADISTICAS MENSUALES:")
# A nivel global
print("Media de los ingresos mensuales:", df_ingresos.sum().mean())
print("Mediana de los ingresos mensuales:", df_ingresos.sum().median())
moda_ingresos = mode(df_ingresos, axis=None).mode
print("Moda de los ingresos mensuales: ", moda_ingresos)
print("Desviacion tipica de los ingresos mensuales:", df_ingresos.sum().std())

print("\nMedia del beneficio mensual:", df_beneficio.sum().mean())
print("Mediana del beneficio mensual:", df_beneficio.sum().median())
moda_beneficio = mode(df_beneficio, axis=None).mode
print("Moda del beneficio mensual: ", moda_beneficio)
print("Desviacion tipica del beneficio mensual:", df_beneficio.sum().std())

# A nivel de cada tienda
print("\nMedia de los ingresos mensuales por tienda:")
print(df_ingresos.mean(axis=1))
print("Mediana de los ingresos mensuales por tienda:")
print(df_ingresos.median(axis=1))
moda_ingresos_tienda = mode(df_ingresos, axis=1).mode
print("Moda de los ingresos mensuales por tienda: ", moda_ingresos_tienda)
print("Desviacion tipica de los ingresos mensuales por tienda:")
print(df_ingresos.std(axis=1))

print("\nMedia del beneficio mensual por tienda:")
print(df_beneficio.mean(axis=1))
print("Mediana del beneficio mensual por tienda:")
print(df_beneficio.median(axis=1))
moda_beneficio_tienda = mode(df_beneficio, axis=1).mode
print("Moda del beneficio mensual por tienda: ", moda_beneficio_tienda)
print("Desviacion tipica del beneficio mensual por tienda:")
print(df_beneficio.std(axis=1))
```

**Explicación**:
- `mean()`: calcula la media del dataframe.
- `median()`: calcula la mediana del dataframe.
- `mode(df_ingresos, axis=None).mode`: obtiene la moda del dataframe "df_ingresos". Para ello, se utiliza una librería de Scikit-learn.
- `std()`: calcula la desviación estándar del dataframe.
