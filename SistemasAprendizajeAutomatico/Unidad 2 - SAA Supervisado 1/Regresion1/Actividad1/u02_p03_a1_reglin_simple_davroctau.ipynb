# REGRESIÓN LINEAL SIMPLE

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv("50_startups.csv")
semilla = 123

X = df[['R&D Spend']]
y = df['Profit']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=semilla)

plt.figure(figsize=(8,4))
sns.scatterplot(x=X_train['R&D Spend'], y=y_train, color='red', label='Train', s=100)
sns.scatterplot(x=X_test['R&D Spend'], y=y_test, color='green', label='Test', s=100)
plt.title('Inversión I+D vs Beneficio de DAVROCTAU', fontsize=14)
plt.xlabel('Inversión I+D', fontsize=12)
plt.ylabel('Beneficio', fontsize=12)
plt.legend()

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

plt.figure(figsize=(8,4))
sns.scatterplot(x=X_train['R&D Spend'], y=y_train, color='red', label='Train', s=100)
sns.scatterplot(x=X_test['R&D Spend'], y=y_test, color='green', label='Test', s=100)
plt.plot(X_test['R&D Spend'], y_pred, color='blue', label='Predicción', linewidth=2)
plt.title('Inversión I+D vs Beneficio de DAVROCTAU', fontsize=14)
plt.xlabel('Gasto en I+D', fontsize=12)
plt.ylabel('Beneficio', fontsize=12)
plt.legend()

plt.show()
```

```python
r2_train = model.score(X_train, y_train)
r2_test = model.score(X_test, y_test)

print(f"Score en train R²: {r2_train}")
print(f"Score en test R²: {r2_test}")
```
