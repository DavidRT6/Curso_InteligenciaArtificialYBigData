## EJERCICIO 1: Elecciones

```python
import matplotlib.pyplot as plt

votosRecibidos = input("Introduce los votos recibidos: ")
votosSeparados = votosRecibidos.split()

partidos = ["Votos PA", "Votos PB", "Votos PC", "Votos PD"]
votosPA, votosPB, votosPC, votosPD = 0, 0, 0, 0

for i in votosSeparados:
    voto = int(i)

    if voto == 1:
        votosPA += 1
    elif voto == 2:
        votosPB += 1
    elif voto == 3:
        votosPC += 1
    elif voto == 4:
        votosPD += 1

votos = [votosPA, votosPB, votosPC, votosPD]
plt.pie(votos, labels=partidos)
plt.show()
```

## EJERCICIO 2: Ventas

```python
import matplotlib.pyplot as plt

rango = int(input("Introduce el rango de años: "))
ventas = []

for i in range(rango):
    numero = input(f"Introduce las ventas del año {i+1}: ")
    numero = int(numero)
    ventas.append(numero)

print(ventas)

plt.plot(ventas)
plt.xlabel("Año")
plt.ylabel("Ventas")
plt.title("Evolución de las Ventas")
plt.show()
```

## EJERCICIO 3: Gráfico circular notas

```python
import matplotlib.pyplot as plt

notas = input("Introduce las notas separadas por espacios: ")
notas = notas.split()
notas = [int(nota) for nota in notas]

categorias = ["Suspensos", "Aprobados", "Notables", "Sobresalientes"]
suspensos, aprobados, notables, sobresalientes = 0,0,0,0

for nota in notas:
    if nota < 5:
        suspensos += 1
    elif nota < 7:
        aprobados += 1
    elif nota < 9:
        notables += 1
    else:
        sobresalientes += 1

valores = [suspensos, aprobados, notables, sobresalientes]

plt.figure(figsize=(8, 8))
plt.pie(
    valores,
    labels=categorias,
    autopct="%1.1f%%",
    colors=["red", "orange", "lightblue", "green"],
    startangle=90,
)
plt.title("Distribución de Notas")
plt.show()
```

## EJERCICIO 5: Múltiples gráficos

Crea un programa MultiplesGraficos.py con un panel 2x2 para mostrar 4 gráficos diferentes sobre el dataset parques_eolicos.csv. En concreto se piden los siguientes gráficos:
- Gráfico circular que muestre la proporción de parques eólicos por provincia
- Gráfico de barras con la potencia media por provincia
- Histograma con la frecuencia del número de aerogeneradores
- Gráfico de cajas que muestre la dispersión del número de aerogeneradores en Valladolid y Palencia

```python
import matplotlib.pyplot as plt
import pandas as pd

titulos = ["Parques por provincia", "Potencia media por provincia", "Número de aerogeneradores", "Aerogeneradores en VA y PA"]

df_lugar = "parques_eolicos.csv"
df = pd.read_csv(df_lugar, encoding="utf-8", sep=";")
df.head()   # Muestra los 5 primeros valores
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 10))
print(df.keys())

# APARTADO 1: Parques por provincia
aerogeneradores_agrupados = df.groupby("provincia")["aerogeneradores"].sum().reset_index()

provincias = aerogeneradores_agrupados.get("provincia", [])
aerogeneradores = aerogeneradores_agrupados.get("aerogeneradores", [])

axes[0,0].pie(
    aerogeneradores,
    labels=provincias,
)
axes[0,0].set_title(titulos[0])
```
